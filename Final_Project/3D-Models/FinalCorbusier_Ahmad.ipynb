```python
final_cells = []
selectors = []
for obj in objects:
    d = Topology.Dictionary(obj)
    name = Dictionary.ValueAtKey(d, "name", "unknown")
    cells = Topology.Cells(obj)
    if len(cells) > 0 :
        for c in cells:
            s = Topology.InternalVertex(c)
            if "kitchen_" in name:
                color = "red"
            elif "dining_" in name:
                color = "green"
            elif "entrance_" in name:
                color = "yellow"
            elif "living_room_" in name:
                color = "orange"
            elif 'bedroom_' in name:
                color = "purple"
            elif 'wc_' in name or 'pantry_' in name:
                color = "cyan"
            elif 'balcony_' in name:
                color = "magenta"
            elif 'corridor_' in name or 'lobby_' in name:
                color = "brown"
            elif 'pantry_' in name:
                color = "pink"
            elif 'core' in name:
                color = "gray"
            d = Dictionary.ByKeysValues(["color", "vertex_size"], [color, 10])
            s = Topology.SetDictionary(s, d)
            selectors.append(s)
            final_cells.append(c)
        print(Dictionary.Keys(d), Dictionary.Values(d))
```

## 7. Creating Cells

```python
hotel = CellComplex.ByCells(final_cells)
hotel = Topology.TransferDictionariesBySelectors(hotel, selectors, tranCells=True)
hotel_cells = Topology.Cells(hotel)
for hotel_cell in hotel_cells:
    d = Topology.Dictionary(hotel_cell)
    print(Dictionary.Keys(d), Dictionary.Values(d))
```

## 11. Get Doors

```python
aperatures = []
with open(r"C:\Users\User\Desktop\00-MaCAD\GraphML_files\Final\doors.corb.obj") as objFile:
    objects = Topology.ByOBJFile(objFile)

# Debug: Check what objects we have
print(f"Total objects: {len(objects)}")

for obj in objects:
    obj_faces = Topology.Faces(obj)
    print(f"Object has {len(obj_faces)} faces")

    # Only process if it has at least one face
    if len(obj_faces) > 0:
        for face in obj_faces:  # Get ALL faces, not just [0]
            face = Topology.RemoveCollinearEdges(face)
            d = Dictionary.ByKeysValues(["type", "color", "vertex_size"], ["door", "orange", 10])
            face = Topology.SetDictionary(face, d)
            aperatures.append(face)
            print(Dictionary.Keys(d), Dictionary.Values(d))

print(f"Total apertures collected: {len(aperatures)}")
```
